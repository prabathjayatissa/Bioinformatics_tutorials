# DNA → Protein BLAST Pipeline
This notebook runs BLASTn and BLASTp, translates DNA to protein, and shows top hits.

```
from Bio.Blast import NCBIWWW, NCBIXML
from Bio.Seq import Seq
```

```
def run_blastn(dna_seq, top_hits=5):
    print("Running BLASTn...")
    result_handle = NCBIWWW.qblast("blastn", "nt", dna_seq)
    blast_records = NCBIXML.parse(result_handle)
    blast_record = next(blast_records)
    hits = []
    for alignment in blast_record.alignments[:top_hits]:
        hsp = alignment.hsps[0]
        hits.append({
            "title": alignment.title,
            "length": alignment.length,
            "score": hsp.score,
            "e_value": hsp.expect,
            "percent_identity": (hsp.identities / hsp.align_length) * 100
        })
    return hits
```

```
def translate_dna(dna_seq):
    dna = Seq(dna_seq)
    return dna.translate()
```

```
def run_blastp(protein_seq, top_hits=5):
    print("Running BLASTp...")
    result_handle = NCBIWWW.qblast("blastp", "nr", protein_seq)
    blast_records = NCBIXML.parse(result_handle)
    blast_record = next(blast_records)
    hits = []
    for alignment in blast_record.alignments[:top_hits]:
        hsp = alignment.hsps[0]
        hits.append({
            "title": alignment.title,
            "length": alignment.length,
            "score": hsp.score,
            "e_value": hsp.expect,
            "percent_identity": (hsp.identities / hsp.align_length) * 100
        })
    return hits
```

```
def print_hits(hits, search_type="BLAST"):
    print(f"\nTop {len(hits)} hits from {search_type}:")
    for i, hit in enumerate(hits, 1):
        print(f"{i}. {hit['title']}")
        print(f"   Length: {hit['length']}  Score: {hit['score']}  E-value: {hit['e_value']:.2e}  Percent identity: {hit['percent_identity']:.2f}%\n")
```

```
# Input DNA sequence
dna_input = input("Enter your DNA sequence: ").strip().upper()

# Run BLASTn
blastn_hits = run_blastn(dna_input)
print_hits(blastn_hits, "BLASTn")

# Translate DNA → Protein
protein_seq = translate_dna(dna_input)
print(f"Translated Protein Sequence:\n{protein_seq}\n")

# Run BLASTp
blastp_hits = run_blastp(protein_seq)
print_hits(blastp_hits, "BLASTp")
```
