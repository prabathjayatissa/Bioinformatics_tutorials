# Sequence Analysis & Alignment Pipeline
This notebook calculates the GC content of two DNA sequences and performs a global pairwise alignment to score their similarity.

```python
from Bio.Seq import Seq
from Bio.SeqUtils import gc_fraction
from Bio import Align
```

```python
def analyze_sequence(name, seq_str):
    """Calculates length and GC content of a sequence."""
    seq = Seq(seq_str)
    gc = gc_fraction(seq) * 100
    print(f"\n--- {name} ---")
    print(f"Length:     {len(seq)} bp")
    print(f"GC Content: {gc:.2f}%")
    return seq
```

```python
def perform_global_alignment(seq1, seq2):
    """Performs a global pairwise alignment using Biopython's modern PairwiseAligner."""
    print("\nRunning Global Alignment...")
    aligner = Align.PairwiseAligner()
    
    # Configure alignment scoring parameters
    aligner.mode = 'global'
    aligner.match_score = 2.0
    aligner.mismatch_score = -1.0
    aligner.open_gap_score = -0.5
    aligner.extend_gap_score = -0.1
    
    # Execute alignment
    alignments = aligner.align(seq1, seq2)
    best_alignment = alignments[0]
    
    print(f"Alignment Score: {best_alignment.score}")
    print("\nBest Alignment Output:")
    print(best_alignment)
    return best_alignment
```

```python
# Input DNA sequences
dna_input_1 = input("Enter your first DNA sequence: ").strip().upper()
dna_input_2 = input("Enter your second DNA sequence to align against: ").strip().upper()

# Analyze basic stats
seq1 = analyze_sequence("Sequence 1", dna_input_1)
seq2 = analyze_sequence("Sequence 2", dna_input_2)

# Run alignment
if dna_input_1 and dna_input_2:
    alignment_result = perform_global_alignment(seq1, seq2)
else:
    print("\nError: Please provide valid sequences for both inputs to run the alignment.")
```
